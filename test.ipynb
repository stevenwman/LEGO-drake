```python
from line_profiler import LineProfiler
from stickbots.stickbot import build_stickbot
from stickbots.geom_builder import generate_feet_geom, _generate_single_foot
from stickbots.config import FeetVars, StickbotParams

# Create a LineProfiler instance
profiler = LineProfiler()

# Add the functions to profile
profiler.add_function(build_stickbot)
profiler.add_function(generate_feet_geom)
profiler.add_function(_generate_single_foot)


# The code to be profiled
def main():
    ft_prm = FeetVars()
    sbp = StickbotParams(feet_vars=ft_prm)
    build_stickbot(sbp)

# Wrap the function call with the profiler
profiler.runcall(main)

# Print the results
profiler.print_stats()
```
